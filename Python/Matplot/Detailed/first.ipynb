```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x=np.linspace(0,5,11)
x
y=x**2
y
```

```python
plt.plot(x,y)
plt.title("Hello")
plt.xlabel("Number")
plt.ylabel("Squares")
```

```python
# subplot
plt.subplot(2,2,1)
plt.plot(x,y)
plt.subplot(2,2,4)
plt.plot(x,y)
plt.subplot(2,2,3)
plt.plot(y,x)
plt.subplot(2,2,2)
plt.plot(x**2,x**2)
```

```python
# object oriented way
fig=plt.figure(figsize=(20,8),dpi=100)
axis1=fig.add_axes([0.1,0.1,1,1]) #shift form left ,right and height and width
axis1.plot(x,y)
axis2=fig.add_axes([0.3,0.4,0.4,0.4])
axis2.plot(x,y)
```

```python
# Types of plot
plt.scatter(x,y)
```
