```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
y=[98,67,88,95,88]
x=["Part1","Part2","Part3","Part4","Part5"]
color=['red','blue','yellow','pink','orange']
plt.bar(x , y,color=color, edgecolor='black')
plt.ylabel("Popularity",fontsize=15)
plt.xlabel("Parts of Harry Potter",fontsize=15)
plt.title("Popularity of different parts of harry potter",fontsize=18)
plt.show()
```

```python
x=["day1","day2","day3","day4","day5"]
y=[300,430,250,230,400]
y1=[500,450,300,250,320]
plt.plot(x,y,marker=".",ls="--",color="red",label="week 1")
plt.plot(x,y1,marker="*",ls="-",color="green",label="week 2",alpha=0.5)
plt.legend()
```

```python
x=np.random.randint(1,10,50)
y=np.random.randint(10,100,50)
color=np.random.randint(10,100,50)
plt.scatter(x,y,marker="*",cmap="hot",c=color)
plt.colorbar()
```

```python
brands=["OnePlus","Apple","Samsung","Nokia","Redmi"]
popularity=[22,35,30,3,10]
c=["red","silver","blue","pink","orange"]
exp=[0,0,0,0,0.2]
plt.pie(popularity,labels=brands,colors=c,explode=exp,shadow=True,autopct="%.2f",startangle=180)
plt.savefig("piechart.png",facecolor="white",pad_inches=2)
```
