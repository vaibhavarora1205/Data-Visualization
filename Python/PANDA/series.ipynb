#### Finding Missing Data

```python
data={
    'A':[1,2,np.nan,4,5],
    'B':[1,2,3,4,5],
    'C':[1,2,3,np.nan,np.nan],
    'D':[1,np.nan,np.nan,np.nan,5],
}
df3=pd.DataFrame(data)
df3
```

```python
df3.isna().sum()
```

```python
df3.isna().any()
```

#### Removing Missing Data

```python
# Works on the basis of Row
df3.dropna()
```

```python
df3.dropna(thresh=3)
```

##### Fill Missing Values with zero

```python
df3.fillna(0)
```

```python
df3
```

#### Fill different values for each column

```python
values={'A':0,'B':100,'C':300,'D':400}
df3.fillna(value=values)
```

Fill With Mean

```python
df3.fillna(df3.mean())
```
