```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Chai Point

# Data structure: [restaurant_id, 2021, 2022, 2023, 2024]
sales_data = np.array([
[1, 150000, 180000, 220000, 250000], # Paradise Biryani
[2, 120000, 140000, 160000, 190000], # Beijing Bites
[3, 200000, 230000, 260000, 300000], # Pizza Hub
[4, 180000, 210000, 240000, 270000], # Burger Point
[5, 160000, 185000, 205000, 230000]
])
```

```python
print("=====Zomato Sales Analysis=====\n")
print("Sales Data Shape",sales_data.shape)
print("Sample Data for first 3 Restaurants: \n",sales_data[:3])
print("Sample Data for first 3 Restaurants: \n",sales_data[:,1:])
print("Sample Data for first 3 Restaurants: \n",sales_data[:,3:])
```

total sales per year

```python
print(np.sum(sales_data,axis=0))
yearly_total=np.sum(sales_data[:,1:],axis=0)
print(yearly_total)
```

minimum sales per restaurant

```python
min_sales=np.min(sales_data[:,1:],axis=1)
print(min_sales)
```

maximum sales per year

```python
max_per_year=np.max(sales_data[:,1:],axis=0)
print(max_per_year)
```

average sales per restaurant

```python
avg_sales=np.mean(sales_data[:,1:],axis=1)
print(avg_sales)
```

Cumulative sales

```python
cumsum=np.cumsum(sales_data[:,1:],axis=1)
print(cumsum)
```

PLOT

```python
plt.figure(figsize=(6,4))
plt.plot(np.mean(cumsum,axis=0))
plt.title("Average sales across all restaurants")
plt.xlabel("Years")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

Vector operations

```python
vector1=np.array([1,2,3,4,5])
vector2=np.array([7,8,9,10,11])
print("Vector Addition: ",vector1+vector2)

print("Vector Multiplication: ",vector1*vector2)
# print("Vector Multiplication: ",vector2*vector1)
print("Dot Product: ",np.dot(vector1,vector2))
angle=np.arccos(np.dot(vector1,vector2)/(np.linalg.norm(vector1)*np.linalg.norm(vector2)))
print("Dot Product: ",np.dot(vector1,vector2))
print(angle)
```
