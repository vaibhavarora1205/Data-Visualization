```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('SuperMarketAnalysis.csv')
df.head()
```

Which city has the highest average transaction value

```python
avgSalesByCity=df.groupby('City')['Sales'].mean()
print(avgSalesByCity.idxmax())
```

Which product line generates the most revenue

```python
revenueByPL=df.groupby('Product line')['gross income'].sum()
print(revenueByPL.idxmax())
```

Which product line has the highest average customer rating

```python
ratingOfPL=df.groupby('Product line')['Rating'].mean()
print(ratingOfPL.idxmax())
```

Do Members or Normal customers have higher average transaction
    values

```python
TransactinsByCustomer=df.groupby('Customer type')['Sales'].mean()
print(TransactinsByCustomer.idxmax())
```

Which payment method is associated with the highest average
    transaction value

```python
TransactinsByPayment=df.groupby('Payment')['Sales'].mean()
print(TransactinsByPayment.idxmax())
```

Which city has the highest total gross income

```python
grossIncomeByCity=df.groupby('City')['gross income'].sum()
print(grossIncomeByCity.idxmax())
```

Which product line sells the highest quantity but doesn’t
    necessarily generate the highest revenue

```python
revenueByPL=df.groupby('Product line')['gross income'].sum()
quantityByPL=df.groupby('Product line')['Quantity'].sum()
print(quantityByPL.idxmax())
```

Which product line sells the highest quantity but doesn’t
    necessarily generate the highest revenue

```python
df.sort_values('Sales')[['City','Product line','Quantity','Sales']].head(5)
```

Find the city + product line combination with the highest total
    revenue.

```python
revenueByCombo = df.groupby(['City', 'Product line'])['gross income'].sum()
revenueByCombo.idxmax()
```

Find the product line with the highest average unit price

```python
df.groupby('Product line')['Unit price'].mean().idxmax()
```

Find the branch with the highest average rating

```python
df.groupby('Branch')['Rating'].mean().idxmax()
```

Create one visualization that summarizes the sales performance of
     all product lines

```python
salesByPL=df.groupby('Product line')['Sales'].sum().reset_index()
salesByPL.plot(kind='bar',x='Product line')
plt.title("Sales Performance by Product lines")
plt.xlabel("Product Lines")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
```

```python
sns.barplot(x='Product line',y='Sales',data=salesByPL,palette='magma')
plt.title("Sales Performance by Product lines")
plt.xlabel("Product Lines")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
```
