```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

```

```python
df=pd.read_csv('SuperMarketAnalysis.csv')
df.head(2)
```

```python
df.groupby('City').size().plot(kind='bar')
plt.show()
```

```python
transactionsByCity=df['City'].value_counts().reset_index()
sns.barplot(x='City',y='count',data=transactionsByCity,palette='rocket')
plt.title("Transactions by City")
```

```python
totalSalesByPL=df.groupby('Product line')['Sales'].sum()
totalSalesByPL
totalSalesByPL.plot(kind="bar",xlabel="Product line",ylabel="Total Sales",title="Total Sales by Product Lines")
```

```python
sns.barplot(totalSalesByPL,palette='flare')
plt.xticks(rotation=45)

# sns.barplot(x='Product line',y='Sales',data=totalSalesByPL)
```

```python
totalSalesByBH=df.groupby('Branch')['Sales'].sum().reset_index()
totalSalesByBH.plot(kind='bar',x='Branch',y='Sales')
```

```python
sns.barplot(x='Branch',y='Sales',data=totalSalesByBH,palette='crest')
plt.title('Total Sales by branch')
```

```python
paymentMethodNumbers=df['Payment'].value_counts()
paymentMethodNumbers.plot(kind='pie',autopct='%1.1f%%',)
# df.head()
```

```python
averageRatingPL=df.groupby('Product line')['Rating'].mean().reset_index()
averageRatingPL.plot(kind='bar')
```

```python
sns.barplot(x='Product line',y='Rating',data=averageRatingPL,palette='vlag')
plt.xticks(rotation=-45)
```

```python
df['Sales'].plot(kind='hist',bins=20,title="Distribution of Sales")
```

```python
df['Rating'].plot(kind='hist',bins=15)
```

```python
df.plot(x='Unit price',y='Sales',kind='scatter')
```

```python
sns.scatterplot(x='Unit price',y='Sales',data=df,marker='*')
```

```python
df['Date'] = pd.to_datetime(df['Date'])
salesByDate = df.groupby('Date')['Sales'].sum()
salesByDate.plot(kind='line')
plt.show()
```
