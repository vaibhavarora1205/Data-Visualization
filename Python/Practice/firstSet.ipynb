#### Feature Extraction

```python
# make a new col for episode count
def extractEpisodes(txt):
    check=False
    data=""
    for i in txt:
        if i==")":
            return data
        if check==True:
            data= data+i
        if i=='(':
            check= True
```

```python
df['Episodes']=df['Title'].apply(extractEpisodes)
df['Episodes']=df['Episodes'].str.replace(" eps","")
df
```

```python
df.loc[0]['Episodes']
df['Episodes']=df['Episodes'].astype(int)
df
```

```python
# Make a new column for time stamps
df.loc[0]['Title']
```

```python
def extractTimeStamps(txt):
    data=""
    for i in range(len(txt)):
        if txt[i]==')':
            for j in range (i+1,i+20):
                data+=txt[j]
            return data

# def extractTimeStamps(txt):
#     check=False
#     data=""
#     for i in txt:
#         if i==",":
#             return data
#         if check==True:
#             data=data+i
#         if i==')':
#             check=True
```

```python
df['Time Stamps']=df['Title'].apply(extractTimeStamps)
# df['Title'].apply(extractTimeStamps)
df
```

```python

def calculate_months(x):
    start, end = x.split(" - ")
    
    start = pd.to_datetime(start, format="%b %Y")
    end = pd.to_datetime(end, format="%b %Y")
    
    return (end.year - start.year) * 12 + (end.month - start.month) + 1

df["Total Months"] = df["Time Stamps"].apply(calculate_months)
```
